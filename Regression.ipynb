2 . What are the key assumptions of Simple Linear Regression ?

1. Linearity
 - There is a linear relationship between the independent variable (X) and the dependent variable (Y).

 - The change in Y is proportional to the change in X.

 - Check: A scatter plot of X vs Y should show a straight-line trend.

2. Independence of Errors
 - The residuals (errors) should be independent of each other.

 - Particularly important in time series data (no autocorrelation).

 - Check: Use the Durbin-Watson test.

3. Homoscedasticity
 - The variance of residuals is the same across all levels of X (i.e., constant spread of errors).

 - Check: Plot residuals vs fitted values — should show no pattern or funnel shape.

4. Normality of Residuals
 - The residuals (errors) should be normally distributed.

 - Check: Use a Q-Q plot or histogram of residuals.

5. No Multicollinearity (Not applicable in Simple Linear Regression)
 - Since there's only one independent variable, this is not a concern here — it applies to multiple regression.

3 . What does the coefficient m represent in the equation Y=mX+c ?

```python
import matplotlib.pyplot as plt
import numpy as np

m = 5
c = 40

X = np.linspace(0, 10, 100)
Y = m * X + c

plt.figure(figsize=(8, 5))
plt.plot(X, Y, label='Y = 5X + 40', linewidth=2)
plt.scatter([2, 4, 6, 8], [5*2+40, 5*4+40, 5*6+40, 5*8+40], color='red', label='Sample Points')
plt.title('Simple Linear Regression Line')
plt.xlabel('Study Hours (X)')
plt.ylabel('Marks (Y)')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

4 . What does the intercept c represent in the equation Y=mX+c ?

```python
# Create a new line with a different intercept
m = 5
c1 = 40  # original intercept
c2 = 20  # new intercept for comparison

# Create X values
X = np.linspace(0, 10, 100)
Y1 = m * X + c1
Y2 = m * X + c2

# Plotting
plt.figure(figsize=(8, 5))
plt.plot(X, Y1, label='Y = 5X + 40', linewidth=2)
plt.plot(X, Y2, label='Y = 5X + 20', linewidth=2, linestyle='--')
plt.title('Effect of Changing the Intercept (c)')
plt.xlabel('Study Hours (X)')
plt.ylabel('Marks (Y)')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

5 . How do we calculate the slope m in Simple Linear Regression ?

```python
import numpy as np

X = np.array([1, 2, 3, 4])
Y = np.array([50, 55, 65, 70])
n = len(X)

sum_X = np.sum(X)
sum_Y = np.sum(Y)
sum_XY = np.sum(X * Y)
sum_X_squared = np.sum(X ** 2)

numerator = n * sum_XY - sum_X * sum_Y
denominator = n * sum_X_squared - sum_X ** 2
m = numerator / denominator

m
```

10 . What are the key assumptions of Multiple Linear Regression ?

1. Linearity
 - The relationship between each independent variable and the dependent variable Y is linear.

 - The combined effect of predictors is additive.

 - Check: Use scatterplots or partial regression plots.

2. Independence of Errors
 - The residuals (errors) are independent from one another.

 - No autocorrelation (especially in time series data).

 - Check: Durbin-Watson test.

3. Homoscedasticity (Constant Variance)
 - The variance of residuals is constant across all levels of independent variables.

 - Check: Plot residuals vs. predicted values — the spread should be even.

4. Normality of Residuals
 - The residuals should be approximately normally distributed.

 - Check: Histogram or Q-Q plot of residuals.

5. No Multicollinearity
 - The independent variables should not be highly correlated with each other.

 - Check: Use Variance Inflation Factor (VIF) — VIF > 10 is a red flag.

6. No Significant Outliers or High Leverage Points
 - Outliers can skew the regression line.

 - High-leverage points have a lot of influence on the model.

 - Check: Cook’s distance, leverage plots.

12 . How can you improve a Multiple Linear Regression model with high multicollinearity ?

1. Ways to Fix or Reduce Multicollinearity
 -  A. Remove Highly Correlated Predictors
    - Drop one of the variables that are highly correlated (e.g., height and arm span)

 -  B. Combine Correlated Variables
    - Combine them into a single feature (e.g., average, ratio, or principal component)

 -  C. Use Principal Component Analysis (PCA)
    - PCA transforms correlated variables into uncorrelated principal components

    - Useful when you want to keep all the information but avoid multicollinearity

 -  D. Use Regularization Techniques
    - Apply Ridge Regression (L2) or Lasso Regression (L1)

    - These shrink coefficients and help manage multicollinearity

 -  E. Collect More Data
    - Sometimes more data helps to distinguish the effects of correlated variables

13 . What are some common techniques for transforming categorical variables for use in regression models ?

1. One-Hot Encoding
 - Converts each category into a new binary column (0 or 1).

 - Best for nominal variables (no order).

2. Label Encoding
 - Assigns a unique integer to each category.

 - Works well only if the variable has natural order (ordinal).

3. Ordinal Encoding
 - Similar to label encoding, but you define the order manually.

4. Binary Encoding / Target Encoding / Frequency Encoding
- These are advanced techniques:

 - Binary Encoding: Uses binary digits for category codes.

 - Target Encoding: Replaces category with mean of target variable.

 - Frequency Encoding: Replaces category with frequency/count.

28 . What are the limitations of polynomial regression ?

1. Risk of Overfitting
 - Higher-degree polynomials can fit the training data too perfectly, including noise.

 - This leads to poor generalization on new data.

 - Example: A 10th-degree polynomial may pass through all points but predict terribly.

2. Poor Extrapolation
 - Predictions outside the range of the training data can become wildly inaccurate.

 - Polynomials tend to explode or oscillate at the ends.

3. Computationally Expensive with Many Variables
 - Adding polynomial and interaction terms in multivariate cases can lead to a huge number of features.

 - Example: With 5 features and degree 3, you might end up with 55+ features.

29 .  What methods can be used to evaluate model fit when selecting the degree of a polynomial ?

1. Visual Inspection (Residual or Prediction Plots)
 - Plot the fitted curve vs. data points

 - Look for patterns in residuals:

    - Random scatter → good fit

    - Curve or funnel shape → underfit or overfit


2. Root Mean Squared Error (RMSE) / Mean Absolute Error (MAE)
 - Measures how far predictions are from actual values

 - Lower RMSE/MAE = better fit

 - Compare these values for different polynomial degrees

4. Cross-Validation (e.g., K-Fold CV)
 - Split the data into training and testing sets multiple times

 - Evaluate performance on unseen data

 - Helps avoid overfitting and gives a reliable estimate of generalization

5. AIC / BIC (Akaike/Bayesian Information Criteria)
 - Used for model selection:

    - Lower AIC/BIC values indicate a better balance between fit and complexity

    - Especially useful in automated or statistical modeling

6. Learning Curves
 - Plot training vs validation error for different polynomial degrees

 - Helps visualize bias vs. variance trade-off



30 . Why is visualization important in polynomial regression ?

1. Reveals the Shape of the Relationship
 - Polynomial regression captures curves (U-shapes, waves, etc.)

 - Plots help you see if the model is truly capturing the trend or just drawing weird curves

2. Detects Underfitting and Overfitting
 - Underfitting → Curve is too simple and misses the pattern

 - Overfitting → Curve is too complex and follows noise

    - Visualization instantly shows whether the chosen degree is too low or too high

3. Highlights Outliers and Influential Points
 - You can spot data points pulling the curve abnormally

 - Outliers in polynomial regression can distort the shape significantly

4. Understands Residual Behavior
 - Plotting residuals vs. predicted values reveals:

 - Random spread (good)

 - Curved/funnel patterns (bad → model may be wrong)

5. Aids in Model Comparison
 - Comparing plots of degree 2 vs. degree 4 polynomial fits helps select the one that balances accuracy and simplicity

6. Improves Communication
 - Visuals make it easy to explain results to non-technical stakeholders

 - A graph of a predicted curve over data is more intuitive than raw metrics

31 . How is polynomial regression implemented in Python ?

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
Y = np.array([1, 4, 9, 16, 25])

poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

model = LinearRegression()
model.fit(X_poly, Y)

Y_pred = model.predict(X_poly)

plt.scatter(X, Y, color='blue', label='Actual data')
plt.plot(X, Y_pred, color='red', label='Polynomial fit')
plt.title('Polynomial Regression (Degree = 2)')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)
plt.show()
```
